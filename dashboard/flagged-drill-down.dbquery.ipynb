```sql
SELECT logical_table_key, table_name, flag_type
FROM (
  SELECT logical_table_key, table_name, 'Low Maturity' AS flag_type
  FROM metadata_governance.gold.table_governance_summary_logical
  WHERE maturity_tier = 'Low'

  UNION ALL

  SELECT logical_table_key, table_name, 'PII Non-Compliant' AS flag_type
  FROM metadata_governance.gold.table_governance_summary_logical
  WHERE pii_non_compliant_count > 0

  UNION ALL

  SELECT logical_table_key, table_name, 'Unowned' AS flag_type
  FROM metadata_governance.gold.table_governance_summary_logical
  WHERE unowned_count > 0

  UNION ALL

  SELECT logical_table_key, table_name, 'Uncertified' AS flag_type
  FROM metadata_governance.gold.table_governance_summary_logical
  WHERE uncertified_count > 0
)
WHERE flag_type = :flag_type
```
