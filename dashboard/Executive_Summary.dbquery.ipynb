```sql
SELECT
  COUNT(*)                                  AS total_tables,
  SUM(total_columns)                        AS total_columns,
  ROUND(AVG(table_completeness_pct), 1)     AS avg_table_completeness_pct,
  SUM(pii_non_compliant_count)              AS total_pii_noncompliant,
  SUM(unowned_count)                        AS total_unowned_columns,
  SUM(uncertified_count)                    AS total_uncertified_columns,
  COUNT(DISTINCT system_name)               AS total_source_systems,
  ROUND(100.0 * (SUM(total_columns) - SUM(unowned_count))     / NULLIF(SUM(total_columns), 0), 1) AS pct_owned,
  ROUND(100.0 * (SUM(total_columns) - SUM(uncertified_count)) / NULLIF(SUM(total_columns), 0), 1) AS pct_certified
FROM metadata_governance.gold.table_governance_summary_logical;
```
